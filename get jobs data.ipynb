```python
countries = [
    "Australia",
    "China",
    "India",
    "Ireland",
    "Israel",
    "Malaysia",
    "New Zealand",
    "Norway",
    "Singapore",
    "Switzerland",
    "Taiwan",
    "Thailand",
    "UK",
    "United Arab Emirates",
    "USA"
]
```

```python
all_jobs = [
    "software engineer",
    "data engineer",
    "bioinformatic"
]
```

```python
# url = "http://oicyhnnf:1pbxqgxqnrc3@"
# proxy_url = [
#     # f"{url}45.38.107.97:6014", 
#     # f"{url}64.137.96.74:6641",
#     # f"{url}198.23.243.226:6361",
#     # f"{url}38.154.185.97:6370",
#     # f"{url}84.247.60.125:6095",
#     # f"{url}191.96.254.138:6185",
#     # f"{url}198.46.161.42:5092",
#     # f"{url}64.137.10.153:5803",
#     # f"{url}104.164.49.38:7693",
#     # f"{url}23.26.71.145:5628",
#     "localhost"
#              ]
for job in all_jobs:
    for country in countries:
        jobs = scrape_jobs(
            site_name=["indeed"], # "bayt", "naukri", "bdjobs", "hellowork"
            search_term=f"'{job}'",
            location=country,
            results_wanted=1000,
            hours_old=7200,
            country_indeed=country,
            # proxies=proxy_url
        )
        print(f"Found {len(jobs)} jobs")
        print(jobs.head())
        jobs.to_csv(
            f"/Workspace/Repos/projects/databricks-batch-pipeline/data/jobs/jobs_{job}_{country}.csv", quoting=csv.QUOTE_NONNUMERIC, escapechar="\\", index=False) # หรือ to_excel
```
