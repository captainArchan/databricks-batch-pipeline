```python
from dataclasses import dataclass, field
from pyspark.sql.functions import *
from pyspark.sql.dataframe import *
from src.common.control_columns import CONTROL_COLUMNS 
```

```python
@dataclass
class GoldLayer:
    pipeline_name: str
    table_name: str
    col: list[str]
    filter_condition: str

    def __post_init__(self) -> None:
        self.table_name_silver = (f"{self.table_name}_silver")


    def load_data(self) -> DataFrame:
        df = (
            spark.table(self.table_name_silver)
            .select(self.col)
            .filter(self.filter_condition)
            .drop(CONTROL_COLUMNS)
              )
        return df


```
