```python
def process_data_to_volumn(
    volume_name: str,
    schema_name: str,
    file_name: str,
    file_type: str
):
    """"""
    create_schema(name=schema_name)
    create_volume(volumn_name=volume_name, schema_name=schema_name)
    if(is_file_in_current_path(file_name=file_name,file_type=file_type)):
        return copy_file_to_volume(
            volume_name = volume_name,
            schema_name = schema_name,
            file_name = file_name,
            file_type = file_type
        )
    else:
        return False
```

```python
def create_schema(name: str):
    """"""
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS workspace.{name}")

```

```python
def create_volume(volumn_name: str, schema_name:str):
    """"""
    spark.sql(f"CREATE VOLUME IF NOT EXISTS {schema_name}.{volumn_name}")
```

```python
import os

def is_file_in_current_path(file_name: str, file_type: str):
    # หา Current Path ปัจจุบันที่ Notebook กำลังรันอยู่
    current_path = os.getcwd()
    
    # นำ Current Path มาต่อกับชื่อไฟล์เพื่อให้ได้ Path เต็ม
    full_path = os.path.join(current_path, "data", f"{file_name}.{file_type}")
    # # เช็คว่ามีไฟล์นี้อยู่จริงหรือไม่ (คืนค่า True/False)
    return os.path.isfile(full_path)
```

```python
def copy_file_to_volume(
        volume_name: str, 
        schema_name: str, 
        file_name: str,
        file_type: str
    ) -> None:
    volume_target = f"/Volumes/workspace/{schema_name}/{volume_name}"
    current_path = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get()
    current_path = "/".join(current_path.split("/")[1:-1])
    
    current_path = f'/Workspace/{current_path}/data/{file_name}.{file_type}'
    if(current_path):
        dbutils.fs.cp(current_path,volume_target)
        return True
    else:
        return False
    # print(f"file {file_name} copied to {volume_target}")
```
